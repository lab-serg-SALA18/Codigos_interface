- **Câmera:** Primeiro, verifique na câmera se:
   - Em `Advanced Camera Settings` clique na opção `Memory Channels` em `1` e depois clique em `Restore`. Assim já vai deixar tudo configurado.

Caso isso não resolva, deixe as configurações assim:
  - Em `Camera Settings` deixe todas as caixas desmarcadas, exceto `Power` e `Absolute Mode`
  - Em `Standard Video Modes` deixe tudo desmarcado
  - Em `Custom Video Modes` deixe o `Pixel Format` em ***Mono 16***, se não fizer isso o código ***quebra!***. Após isso em `Image` deixe a **Width e Height** em 24, ajuste o **Left e Top** para posicionar o NV no centro. Do jeito que eu deixei estava em `Left = 278, Top = 266`. Por fim, clique em `Apply`
  - Em `Camera Registers` não mexa em nada (seria útil para FPGAs, no nosso caso não faz nada)
  - Em `Trigger / Strobe` deixe o `Trigger Delay` desmarcado, `Pin Direction Control` somente com o **GPIO 1** no **Out** (os outros no In), `Strobe Control` desmarque todos.
  O programa mexe em `Trigger Control, Trigger Souce e Trigger Polarity` automaticamente, não precisa mudar nada nesses
  - Em `Advanced Camera Settings` `Y16 Endianness` marque a opção **Little endian (PRG-specific Y16 mode)**, `Embedded Image Information` deixe todos desmarcados. O resto não precisa mexer
  - Em `High Dynamic Range` deixe **desmarcado**
  - Em `Lool Up Table` deixe **desmarcado**
  - Em `Frame Buffer` deixe **desmarcado**
  - Em `Data Flash` não mexa em nada
  - Em `Bus Topology` não mexa em nada
